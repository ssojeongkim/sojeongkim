```python
#NumPy 배열 연산 : 유니버셜 함수
#The Slowness of Loops
#루프는 느리다.
```

```python
import numpy as np
np.random.seed(0)

def compute_reciprocals(values):
    output = np.empty(len(values))
    for i in range(len(values)):
        output[i] = 1.0 / values[i]
    return output
        
values = np.random.randint(1, 10, size=5)
compute_reciprocals(values)
```

```python
big_array = np.random.randint(1, 100, size=1000000)
%timeit compute_reciprocals(big_array)
```

```python
#Centering an array 배열을 중앙으로 정렬하기
X = np.random.random((10, 3))
```

```python
Xmean = X.mean(0)
Xmean 
# mean 집계함수를 사용하여 첫번째 차원의 특성별 평균값을 계산할 수 있다.
# X 배열의 평균을 구하는데, axis=0 옵션은 "열(column) 방향으로" 평균을 계산하라는 뜻입니다. 
#즉, 각 열의 평균값을 따로따로 구합니다.
```

```python
X_centered = X - Xmean
#평균값을 뺌으로써 X배열을 중앙 정렬할 수 있다.
```

```python
import numpy as np

def selection_sort(x):
    for i in range(len(x)):
        swap = i + np.argmin(x[i:])
        (x[i], x[swap]) = (x[swap], x[i])
    return x
```

```python
x = np.array([2, 1, 4, 3, 5]) 
selection_sort(x) #선택 정렬
```

```python
def bogosort(x):
    while np.any(x[:-1] > x[1:]):
        np.random.shuffle(x)
    return x
```

```python
x = np.array([2, 1, 4, 3, 5])
bogosort(x) #보고 정렬(고전적인 정렬 알고리즘) 
```

```python
X = rand.rand(10, 2) #우선 2차원 평면에 임의의 점 10개를 가지는 집합을 만들기. 10*2배열에 배치
```

```python
import matplotlib.pyplot as plt
import seaborn; seaborn.set() # 플롯 스타일링
plt.scatter(X[:, 0], X[:, 1], s=100)
```

```python
dist_sq = np.sum((X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2, axis=-1)
#각 쌍의 점 사이의 거리를 계산 
```

```python
# 각 쌍의 점들 사이의 좌표 차이를 계산함
differences = X[:, np.newaxis, :] - X[np.newaxis, :, :]
differences.shape
```

```python
# 좌표 차이를 제곱함
sq_differences = differences ** 2
sq_differences.shape
```

```python
# 제곱 거리를 구하기 위해 좌표 차이를 더함
dist_sq = sq_differences.sum(-1)
dist_sq.shape
```

```python
dist_sq.diagonal()
```

```python
nearest = np.argsort(dist_sq, axis=1)
print(nearest)
#제곱거리를 변환하면 np.argsort를 이용해 행별로 정렬할 수 있다.
#그러면 가장 완쪽 열이 가장 가까운 이웃의 인덱스를 제공할 것이다.
```

```python
K = 2
nearest_partition = np.argpartition(dist_sq, K + 1, axis=1)
```

```python
plt.scatter(X[:, 0], X[:, 1], s=100)

# 각 점에서 두 개의 가장 가까운 이웃과 선으로 이음
K = 2

for i in range(X.shape[0]):
    for j in nearest_partition[i, :K+1]:
        # X[i]부터 X[j]까지 선으로 이음
        # zip 매직 함수를 이용함
        plt.plot(*zip(X[j], X[i]), color='black')
```
